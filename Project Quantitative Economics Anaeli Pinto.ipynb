## Basic Solow Model

The model begins with the aggregate production function:

$$
Y_t = F(K_t, L_t)
$$

A common specification is the Cobb Douglas function:

$$
Y_t = K_t^\alpha L_t^{1-\alpha}
$$

with \( 0 < \alpha < 1 \), which implies diminishing returns to capital.

We express the model in per worker terms:

$$
y_t = \frac{Y_t}{L_t}, \quad k_t = \frac{K_t}{L_t}
$$

which gives:

$$
y_t = k_t^\alpha
$$

The law of motion of capital is:

$$
k_{t+1} = s k_t^\alpha + (1-\delta)k_t
$$

The steady state occurs when capital does not change over time:

$$
k_{t+1} = k_t
$$

which implies:

$$
k_t = k^*
$$

## Steady State of the Basic Model

At steady state:

$$
k^* = s (k^*)^\alpha + (1-\delta)k^*
$$

Rearranging:

$$
s (k^*)^\alpha = \delta k^*
$$

Solving for \(k^*\):

$$
k^* = \left(\frac{s}{\delta}\right)^{\frac{1}{1-\alpha}}
$$

Output at steady state:

$$
y^* = (k^*)^\alpha
$$

## Comparative Statics (Basic Model)

The steady state level of capital per worker depends on key parameters of the model:

- A higher savings rate $s$ increases the steady state level of capital, $k^*$
- A higher depreciation rate $\delta$ decreases $k^*$
- A higher capital share $\alpha$ affects the curvature of the production function and therefore changes the equilibrium level of $k^*$

These relationships can be derived analytically from the steady state expression:

$$
k^* = \left(\frac{s}{\delta}\right)^{\frac{1}{1-\alpha}}
$$

From this expression, we can clearly observe how each parameter influences the equilibrium. These effects will be illustrated graphically and through simulations.

## Extended Model with Technological Progress and Population Growth

We extend the model by introducing technology \(A_t\) and population growth.

The production function becomes:

$$
Y_t = K_t^\alpha (A_t L_t)^{1-\alpha}
$$

Define variables per effective worker:

$$
\tilde{k}_t = \frac{K_t}{A_t L_t}, \quad \tilde{y}_t = \frac{Y_t}{A_t L_t}
$$

Then:

$$
\tilde{y}_t = \tilde{k}_t^\alpha
$$

Assume:

$$
A_{t+1} = (1+g)A_t, \quad L_{t+1} = (1+n)L_t
$$

The law of motion becomes:

$$
\tilde{k}_{t+1} = \frac{s \tilde{k}_t^\alpha + (1-\delta)\tilde{k}_t}{(1+g)(1+n)}
$$

## Comparative Statics (Extended Model)

In the extended model:

- Higher savings increases steady state capital
- Higher depreciation reduces capital
- Higher population growth \(n\) reduces capital per effective worker
- Higher technological growth \(g\) also reduces capital per effective worker

These effects will be explored using graphs and interactive simulations.

## Interactive Analysis

To better understand the behavior of the model, interactive graphs will be used.

Using sliders for parameters such as:

- savings rate \(s\)
- depreciation \(\delta\)
- population growth \(n\)
- technological growth \(g\)

we can observe how the steady state changes dynamically.

This allows for a more intuitive understanding of comparative statics.

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

sp.init_printing()
```

```python
# Symbols we are going to use
k, s, delta, alpha = sp.symbols('k s delta alpha', positive=True)

# Production function
y = k**alpha

# Law of motion
k_next = s * k**alpha + (1 - delta) * k

print("Production function:")
display(sp.Eq(sp.Symbol('y'), y))

print("Law of motion of capital:")
display(sp.Eq(sp.Symbol('k_{t+1}'), k_next))
```

```python
# Steady state condition
steady_state_eq = sp.Eq(k, s * k**alpha + (1 - delta) * k)

print("Steady state condition:")
display(steady_state_eq)

# Rearranged version of steady state
steady_state_rearranged = sp.Eq(s * k**alpha, delta * k)

print("Rearranged steady state condition:")
display(steady_state_rearranged)

# Solve for k^star
k_star = (s / delta)**(1 / (1 - alpha))

print("Steady state capital per worker:")
display(sp.Eq(sp.Symbol('k^*'), k_star))

# Steady state output
y_star = sp.simplify(k_star**alpha)

print("Steady state output per worker:")
display(sp.Eq(sp.Symbol('y^*'), y_star))
```

```python
# Random parameter values
s_val = 0.25
delta_val = 0.05
alpha_val = 0.3

# Input the values for evaluating the model
k_star_num = float(k_star.subs({s: s_val, delta: delta_val, alpha: alpha_val}))
y_star_num = float(y_star.subs({s: s_val, delta: delta_val, alpha: alpha_val}))

print(f"Using s = {s_val}, delta = {delta_val}, alpha = {alpha_val}")
print(f"Steady state capital per worker (k*): {k_star_num:.4f}")
print(f"Steady state output per worker (y*): {y_star_num:.4f}")
```

```python
# Functions for plotting
def investment(k, s, alpha):
    return s * (k ** alpha)

def break_even(k, delta):
    return delta * k

# Grid for capital
k_grid = np.linspace(0.01, 20, 400)

# Values
inv_vals = investment(k_grid, s_val, alpha_val)
dep_vals = break_even(k_grid, delta_val)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(k_grid, inv_vals, label=r'Investment: $s k^\alpha$')
plt.plot(k_grid, dep_vals, label=r'Break-even: $\delta k$')

# Steady state point
plt.scatter(k_star_num, investment(k_star_num, s_val, alpha_val), s=70)
plt.axvline(k_star_num, linestyle='--', alpha=0.7)
plt.axhline(investment(k_star_num, s_val, alpha_val), linestyle='--', alpha=0.7)

plt.title("Basic Solow Model")
plt.xlabel("Capital per worker, k")
plt.ylabel("Output / Investment")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
def capital_path(k0, s, delta, alpha, T=40):
    path = [k0]
    for _ in range(T):
        k_next = s * (path[-1] ** alpha) + (1 - delta) * path[-1]
        path.append(k_next)
    return np.array(path)

k0 = 1.0
T = 40
path = capital_path(k0, s_val, delta_val, alpha_val, T)

# Plot path
plt.figure(figsize=(10, 6))
plt.plot(range(T + 1), path, marker='o', markersize=3)
plt.axhline(k_star_num, linestyle='--', label=r'Steady state $k^*$')
plt.title("Convergence of Capital per Worker")
plt.xlabel("Time")
plt.ylabel("Capital per worker")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
s_values = [0.15, 0.25, 0.35]

plt.figure(figsize=(10, 6))

for s_temp in s_values:
    inv_vals = investment(k_grid, s_temp, alpha_val)
    k_star_temp = float(((s_temp / delta_val) ** (1 / (1 - alpha_val))))
    plt.plot(k_grid, inv_vals, label=fr'$s = {s_temp}$')
    plt.scatter(k_star_temp, investment(k_star_temp, s_temp, alpha_val), s=50)

plt.plot(k_grid, break_even(k_grid, delta_val), label=r'$\delta k$', linewidth=2)

plt.title("Comparative Statics: Effect of Savings Rate on Steady State")
plt.xlabel("Capital per worker, k")
plt.ylabel("Investment / Break-even")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
def interactive_solow(s=0.25, delta=0.05, alpha=0.3):
    k_grid = np.linspace(0.01, 20, 400)
    inv_vals = s * (k_grid ** alpha)
    dep_vals = delta * k_grid
    k_star_val = (s / delta) ** (1 / (1 - alpha))
    y_star_val = k_star_val ** alpha

    plt.figure(figsize=(10, 6))
    plt.plot(k_grid, inv_vals, label=r'Investment: $s k^\alpha$')
    plt.plot(k_grid, dep_vals, label=r'Break-even: $\delta k$')
    plt.scatter(k_star_val, s * (k_star_val ** alpha), s=70, label=fr'Steady state $k^*={k_star_val:.2f}$')
    plt.axvline(k_star_val, linestyle='--', alpha=0.7)
    plt.title("Interactive Basic Solow Model")
    plt.xlabel("Capital per worker, k")
    plt.ylabel("Investment / Break-even")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

    print(f"Steady state capital per worker (k*): {k_star_val:.4f}")
    print(f"Steady state output per worker (y*): {y_star_val:.4f}")

interact(
    interactive_solow,
    s=FloatSlider(value=0.25, min=0.05, max=0.60, step=0.01),
    delta=FloatSlider(value=0.05, min=0.01, max=0.20, step=0.01),
    alpha=FloatSlider(value=0.30, min=0.10, max=0.90, step=0.05)
);
```

## Interpretation of the Basic Model

According to the results, the economic system reaches an equilibrium point at which there are no net wealth changes because capital investment equals the amount that has been lost through depreciation of capital. This is represented graphically by the intersection of an investment curve and a depreciation line.

The results from comparing the stats confirm the assumptions of the Solow growth model. Increasing the savings rate raises the investment curve, shifting it up in the graph and increasing the steady-state capital level. Conversely, increasing the depreciation rate reduces both the amount of capital that can be maintained and the steady-state capital level. Both parameters are causal to the shape of the production function and thus affect the equilibrium point.

```python
# Added symbols to solve the extended model
k_tilde, n, g = sp.symbols('k_tilde n g', positive=True)

# Production per effective worker
y_tilde = k_tilde**alpha

# Law of motion for effectice worker
k_tilde_next = (s * k_tilde**alpha + (1 - delta) * k_tilde) / ((1 + n) * (1 + g))

print("Production per effective worker:")
display(sp.Eq(sp.Symbol(r'\tilde{y}'), y_tilde))

print("Law of motion of capital per effective worker:")
display(sp.Eq(sp.Symbol(r'\tilde{k}_{t+1}'), k_tilde_next))
```

```python
# Steady state condition
steady_state_ext = sp.Eq(k_tilde, (s * k_tilde**alpha + (1 - delta) * k_tilde) / ((1 + n) * (1 + g)))

print("Extended model steady state condition:")
display(steady_state_ext)

# Std approximation used in Solow model
k_tilde_star = (s / (delta + n + g))**(1 / (1 - alpha))
y_tilde_star = sp.simplify(k_tilde_star**alpha)

print("Steady state capital per effective worker:")
display(sp.Eq(sp.Symbol(r'\tilde{k}^*'), k_tilde_star))

print("Steady state output per effective worker:")
display(sp.Eq(sp.Symbol(r'\tilde{y}^*'), y_tilde_star))
```

```python
# Parameter values
s_val = 0.25
delta_val = 0.05
alpha_val = 0.30
n_val = 0.02
g_val = 0.03

# Numerical steady state values
k_tilde_star_num = float(k_tilde_star.subs({
    s: s_val,
    delta: delta_val,
    alpha: alpha_val,
    n: n_val,
    g: g_val
}))

y_tilde_star_num = float(y_tilde_star.subs({
    s: s_val,
    delta: delta_val,
    alpha: alpha_val,
    n: n_val,
    g: g_val
}))

print(f"Using s = {s_val}, delta = {delta_val}, alpha = {alpha_val}, n = {n_val}, g = {g_val}")
print(f"Steady state capital per effective worker (k̃*): {k_tilde_star_num:.4f}")
print(f"Steady state output per effective worker (ỹ*): {y_tilde_star_num:.4f}")
```

```python
def investment_ext(k, s, alpha):
    return s * (k ** alpha)

def break_even_ext(k, delta, n, g):
    return (delta + n + g) * k

# Grid
k_grid_ext = np.linspace(0.01, 20, 400)

# Values
inv_ext_vals = investment_ext(k_grid_ext, s_val, alpha_val)
be_ext_vals = break_even_ext(k_grid_ext, delta_val, n_val, g_val)

plt.figure(figsize=(10, 6))

plt.plot(k_grid_ext, inv_ext_vals, label=r'Investment: $s \tilde{k}^{\alpha}$', linewidth=2)
plt.plot(k_grid_ext, be_ext_vals, label=r'Break-even: $(\delta+n+g)\tilde{k}$', linewidth=2)

# Steady state point
plt.scatter(
    k_tilde_star_num,
    investment_ext(k_tilde_star_num, s_val, alpha_val),
    color='black',
    s=80,
    zorder=5
)

plt.axvline(k_tilde_star_num, linestyle='--', alpha=0.7)
plt.axhline(investment_ext(k_tilde_star_num, s_val, alpha_val), linestyle='--', alpha=0.7)

plt.text(
    k_tilde_star_num + 0.4,
    investment_ext(k_tilde_star_num, s_val, alpha_val),
    r'$\tilde{k}^*$',
    fontsize=12
)

plt.title("Extended Solow Model: Steady State per Effective Worker", fontsize=14)
plt.xlabel(r"Capital per effective worker, $\tilde{k}$", fontsize=12)
plt.ylabel("Output / Investment", fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Graph Interpretation of the Extended Model

In the extended Solow model, the steady state is determined by the intersection between the investment curve $s \tilde{k}^\alpha$ and the break-even line $(\delta + n + g)\tilde{k}$.

Compared to the basic model, the break-even line is steeper because the economy must now devote resources not only to replacing depreciated capital, but also to equip a growing population and keep up with technological progress.

As a result, higher values of $n$ or $g$ reduce the steady state level of capital per effective worker.

```python
def capital_path_extended(k0, s, delta, alpha, n, g, T=40):
    path = [k0]
    for _ in range(T):
        k_next = (s * (path[-1] ** alpha) + (1 - delta) * path[-1]) / ((1 + n) * (1 + g))
        path.append(k_next)
    return np.array(path)

# Simulate path
k0_ext = 1.0
T = 40
path_ext = capital_path_extended(k0_ext, s_val, delta_val, alpha_val, n_val, g_val, T)

plt.figure(figsize=(10, 6))
plt.plot(range(T + 1), path_ext, marker='o', markersize=3)
plt.axhline(k_tilde_star_num, linestyle='--', label=r'Steady state $\tilde{k}^*$')
plt.title("Convergence of Capital per Effective Worker")
plt.xlabel("Time")
plt.ylabel(r"Capital per effective worker, $\tilde{k}$")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
def interactive_extended_solow(s=0.25, delta=0.05, alpha=0.3, n=0.02, g=0.03):
    k_grid = np.linspace(0.01, 20, 400)
    investment_vals = s * (k_grid ** alpha)
    break_even_vals = (delta + n + g) * k_grid
    k_star_val = (s / (delta + n + g)) ** (1 / (1 - alpha))
    y_star_val = k_star_val ** alpha

    plt.figure(figsize=(10, 6))
    plt.plot(k_grid, investment_vals, label=r'Investment: $s\tilde{k}^{\alpha}$', linewidth=2)
    plt.plot(k_grid, break_even_vals, label=r'Break-even: $(\delta+n+g)\tilde{k}$', linewidth=2)

    plt.scatter(
        k_star_val,
        s * (k_star_val ** alpha),
        color='black',
        s=80,
        zorder=5,
        label=fr'Steady state $\tilde{{k}}^*={k_star_val:.2f}$'
    )

    plt.axvline(k_star_val, linestyle='--', alpha=0.7)
    plt.axhline(s * (k_star_val ** alpha), linestyle='--', alpha=0.7)

    plt.title("Interactive Extended Solow Model", fontsize=14)
    plt.xlabel(r"Capital per effective worker, $\tilde{k}$", fontsize=12)
    plt.ylabel("Output / Investment", fontsize=12)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

    print(f"Steady state capital per effective worker (k̃*): {k_star_val:.4f}")
    print(f"Steady state output per effective worker (ỹ*): {y_star_val:.4f}")

interact(
    interactive_extended_solow,
    s=FloatSlider(value=0.25, min=0.05, max=0.60, step=0.01),
    delta=FloatSlider(value=0.05, min=0.01, max=0.20, step=0.01),
    alpha=FloatSlider(value=0.30, min=0.10, max=0.90, step=0.05),
    n=FloatSlider(value=0.02, min=0.00, max=0.10, step=0.01),
    g=FloatSlider(value=0.03, min=0.00, max=0.10, step=0.01)
);
```

## Interpretation of the Extended Model

In the extended Solow model, the economy converges to a steady state of capital per effective worker. In this framework, the break-even level of investment is equal to depreciation plus population growth plus technological growth.

The extension of the Solow model shows that higher savings rates raise the steady state of capital per effective worker; conversely, higher rates of depreciation, higher population growth, or higher rates of technological advancement lower the steady state level.

The interactive graph demonstrates how the movements in the parameter space shift the curves and therefore change the equilibrium.

## Comparison Between the Basic and Extended Models

The basic Solow model and the extended Solow model share the same fundamental structure, but differ in how they account for long run growth.

In the basic model, the steady state is determined by:

$$
k^* = \left(\frac{s}{\delta}\right)^{\frac{1}{1-\alpha}}
$$

Under this framework, capital accumulation can be said to drive the entire economy towards a stable state represented by constant levels of output per worker in time. When the economy reaches this point, there will be no overall expansion in output per worker going forward.

The extended model incorporates both population increase through time and technological advancement. Capital per effective worker reaches a stable level as defined by:

$$
\tilde{k}^* = \left(\frac{s}{\delta + n + g}\right)^{\frac{1}{1-\alpha}}
$$

The addition of $n$ and $g$ adds two critical changes to the model's dynamics.

First, there is now a higher requirement for invested capital to break even because the new capital has to replace the old capital that was depreciated, increase due to the need to support a larger population, and provide for the changes in technology needed to keep the economy growing. Therefore, in the extended model, the long-run steady-state capital per effective worker level will be lower than in the basic version of the model.

Second, technological progress leads to a new source of long-run economic growth. In the extended model, capital per effective worker reaches a steady-state level; however, total output and output per worker can still rise over the long term as productivity improves.

Consequently, the extended version of this economic growth model allows for a more realistic portrayal of the mechanisms driving economic growth by taking into account the impact of demographic change and technological change as essential to real economies.
